```python
import tkinter as tk
import speech_recognition as sr


# +
# 1. BASIC SETTINGS
# ==========================================+++

LOCATION = "Hyderabad"
MAPS_LINK = (
    "https://www.google.com/maps/search/?api=1&query=Hyderabad"
)


# ============================================================
# 2. MAIN APPLICATION WINDOW
# ============================================================

app = tk.Tk()

app.title("Women Safety AI System")
app.geometry("600x800")
app.configure(bg="white")


# ============================================================
# 3. TITLE
# ============================================================

title_label = tk.Label(
    app,
    text="WOMEN SAFETY AI SYSTEM",
    font=("Arial", 24, "bold"),
    bg="white",
    fg="black"
)

title_label.pack(pady=(25, 5))


subtitle_label = tk.Label(
    app,
    text="AI-Based Emergency Protection System",
    font=("Arial", 12),
    bg="white",
    fg="gray"
)

subtitle_label.pack(pady=(0, 20))


# ============================================================
# 4. SAFETY STATUS
# ============================================================

status_label = tk.Label(
    app,
    text="● STATUS: SAFE",
    font=("Arial", 17, "bold"),
    bg="white",
    fg="green"
)

status_label.pack(pady=10)


# ============================================================
# 5. EMERGENCY CONTACT
# ============================================================

contact_title = tk.Label(
    app,
    text="Emergency Contact",
    font=("Arial", 13, "bold"),
    bg="white"
)

contact_title.pack(pady=(15, 5))


contact_entry = tk.Entry(
    app,
    font=("Arial", 13),
    width=28,
    justify="center"
)

contact_entry.pack(pady=5)


contact_status = tk.Label(
    app,
    text="Enter emergency contact number",
    font=("Arial", 9),
    bg="white",
    fg="gray"
)

contact_status.pack(pady=3)


def save_contact():

    contact = contact_entry.get().strip()

    if contact:

        contact_status.config(
            text="✓ Emergency contact saved",
            fg="green"
        )

    else:

        contact_status.config(
            text="Please enter a contact number",
            fg="red"
        )


save_contact_button = tk.Button(
    app,
    text="SAVE CONTACT",
    font=("Arial", 11, "bold"),
    command=save_contact,
    width=18
)

save_contact_button.pack(pady=5)


# ============================================================
# 6. LOCATION DISPLAY
# ============================================================

location_label = tk.Label(
    app,
    text="📍 Location: Not detected",
    font=("Arial", 13, "bold"),
    bg="white",
    fg="black"
)

location_label.pack(pady=(15, 5))


# ============================================================
# 7. ALERT DISPLAY AREA
# ============================================================

alert_frame = tk.Frame(
    app,
    bg="white",
    bd=2,
    relief="groove"
)

alert_frame.pack(
    padx=40,
    pady=15,
    fill="x"
)


alert_label = tk.Label(
    alert_frame,
    text="Emergency alerts will appear here",
    font=("Arial", 11),
    bg="white",
    fg="black",
    justify="left",
    wraplength=480,
    padx=15,
    pady=15
)

alert_label.pack()


# ============================================================
# 8. SOS EMERGENCY FUNCTION
# ============================================================

def sos_emergency():

    contact = contact_entry.get().strip()

    if contact == "":
        contact = "Not provided"

    status_label.config(
        text="● EMERGENCY DETECTED!",
        fg="red"
    )

    alert_label.config(
        text=(
            "🚨 EMERGENCY ALERT\n\n"
            "Emergency Contact: " + contact +
            "\n\n"
            "📍 Location: " + LOCATION +
            "\n\n"
            "🗺️ Google Maps:\n" + MAPS_LINK +
            "\n\n"
            "⚠️ Please provide immediate assistance."
        ),
        fg="red"
    )

    print("================================")
    print("EMERGENCY ALERT")
    print("Emergency Contact:", contact)
    print("Location:", LOCATION)
    print("Google Maps:", MAPS_LINK)
    print("Please provide immediate assistance.")
    print("================================")


# ============================================================
# 9. VOICE / SPEECH RECOGNITION FUNCTION
# ============================================================

def voice_emergency():

    recognizer = sr.Recognizer()

    try:

        status_label.config(
            text="● LISTENING...",
            fg="blue"
        )

        alert_label.config(
            text="🎤 Listening... Please speak now.",
            fg="blue"
        )

        app.update()

        with sr.Microphone() as source:

            recognizer.adjust_for_ambient_noise(
                source,
                duration=1
            )

            audio = recognizer.listen(
                source,
                timeout=10,
                phrase_time_limit=6
            )

        recognized_text = recognizer.recognize_google(audio)

        print("Recognized speech:", recognized_text)

        emergency_words = [
            "help",
            "emergency",
            "danger",
            "save me",
            "help me"
        ]

        speech = recognized_text.lower()

        emergency_found = any(
            word in speech
            for word in emergency_words
        )

        if emergency_found:

            sos_emergency()

        else:

            status_label.config(
                text="● STATUS: SAFE",
                fg="green"
            )

            alert_label.config(
                text=(
                    "Voice detected:\n"
                    + recognized_text +
                    "\n\n"
                    "No emergency keyword detected."
                ),
                fg="black"
            )

    except sr.UnknownValueError:

        status_label.config(
            text="● VOICE NOT UNDERSTOOD",
            fg="orange"
        )

        alert_label.config(
            text=(
                "Could not understand the voice.\n\n"
                "Please speak clearly and try again."
            ),
            fg="orange"
        )

    except sr.RequestError:

        status_label.config(
            text="● VOICE SERVICE ERROR",
            fg="red"
        )

        alert_label.config(
            text=(
                "Speech recognition service is unavailable."
            ),
            fg="red"
        )

    except Exception as error:

        status_label.config(
            text="● MICROPHONE ERROR",
            fg="red"
        )

        alert_label.config(
            text=(
                "Could not access the microphone.\n\n"
                "Error: " + str(error)
            ),
            fg="red"
        )

        print("Microphone error:", error)


# ============================================================
# 10. GET LOCATION FUNCTION
# ============================================================

def get_location():

    location_label.config(
        text="📍 Location: " + LOCATION,
        fg="black"
    )

    alert_label.config(
        text=(
            "CURRENT LOCATION\n\n"
            "📍 " + LOCATION +
            "\n\n"
            "🗺️ Google Maps:\n"
            + MAPS_LINK
        ),
        fg="black"
    )

    print("Current location:", LOCATION)
    print("Google Maps:", MAPS_LINK)


# ============================================================
# 11. SOS BUTTON
# ============================================================

sos_button = tk.Button(
    app,
    text="🚨 SOS EMERGENCY",
    font=("Arial", 14, "bold"),
    command=sos_emergency,
    width=24,
    height=2
)

sos_button.pack(pady=8)


# ============================================================
# 12. VOICE EMERGENCY BUTTON
# ============================================================

voice_button = tk.Button(
    app,
    text="🎤 VOICE EMERGENCY",
    font=("Arial", 14, "bold"),
    command=voice_emergency,
    width=24,
    height=2
)

voice_button.pack(pady=8)


# ============================================================
# 13. GET LOCATION BUTTON
# ============================================================

location_button = tk.Button(
    app,
    text="📍 GET LOCATION",
    font=("Arial", 14, "bold"),
    command=get_location,
    width=24,
    height=2
)

location_button.pack(pady=8)


# ============================================================
# 14. START APPLICATION
# ============================================================

app.mainloop()
```
